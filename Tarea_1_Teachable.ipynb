# Práctica: Clasificador de Imágenes con Teachable Machine

## 1. Introducción
El objetivo de esta práctica es entrenar un modelo de inteligencia artificial para reconocer imágenes con Teachable Machine y ejecutarlo en una web en local con TensorFlow.js.

Mejoras añadidas a la web:
- Diseño responsive y centrado.
- Botones para iniciar y parar la cámara.
- Selector de umbral mínimo de acierto.
- Mensaje personalizado con el resultado.
- Subida de fotos desde el ordenador.
- Historial que guarda las detecciones.



## 2. Entrenamiento del modelo
He creado dos clases: **Mano** y **Ratón**, subiendo más de 20 fotos para que el modelo aprenda a diferenciarlas bien.

## 3. Exportación
- Exporté y descargué el modelo en formato **Tensorflow.js** para usarlo en la web (`model.json`, `metadata.json` y `weights.bin`).
- También lo exporté en formato **TensorFlow** para guardarlo para futuras prácticas.

## 4. Pruebas de funcionamiento

### Prueba 1: Foto propia (Mano al 100%)
Probando una foto tomada en el escritorio, el modelo la clasifica correctamente con un 100% de confianza y la guarda en el historial.

## 5. Conclusión
El modelo funciona de forma rápida en el navegador gracias a TensorFlow.js. La clave para que no falle al subir fotos es que las imágenes de entrenamiento tengan distintos ángulos y que el fondo no engañe a la red neuronal.
