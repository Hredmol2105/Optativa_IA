**¿Cómo se usa el modelo exportado? ¿Cómo se comporta?**

---



**¿Cómo se usa el modelo exportado?**

Carga del modelo: Se cargan los pesos del modelo (keras_model.h5) usando tf_keras con compatibilidad de Keras clásica (compile=False), ya que los modelos exportados por Teachable Machine requieren la arquitectura de Keras 2.

Carga de etiquetas: Se lee el fichero labels.txt para mapear los índices numéricos de salida con el nombre de cada clase.

Preprocesamiento de la imagen:
Se abre la imagen y se fuerza a espacio de color RGB.
Se reescala y recorta al centro a un tamaño fijo de 224 × 224 píxeles con interpolación Lanczos.
Se convierte a un array NumPy de tipo float32 y se normaliza su rango de color al intervalo $[-1, 1]$ aplicando la fórmula (pixel / 127.5)

Inferencia: Se pasa el tensor resultante con forma (1, 224, 224, 3) por model.predict(), extrayendo la clase con mayor probabilidad mediante np.argmax() y su puntuación de confianza asociada.




**¿Cómo se comporta el modelo?**

Con la imagen clara: El modelo clasifica el objeto de forma correcta y con un Confidence Score muy alto (cercano a 0.90 – 1.00), ya que las características visuales coinciden directamente con la distribución de los datos de entrenamiento.   

Con la imagen ambigua: El nivel de confianza disminuye notablemente o se reparte de forma inestable entre las clases disponibles. Al presentarse elementos solapados, distorsionados o con cambios de iluminación/ángulo, la red duda y puede cometer falsos positivos según qué rasgos dominen en el centro de la imagen.   

Con la imagen sin categoría (fuera de dominio): El modelo no tiene noción de "desconocido" o "ninguna de las anteriores", ya que utiliza una capa final Softmax que obliga matemáticamente a que la suma de las probabilidades de todas las clases sea siempre igual a 1 (100%). Por tanto, clasifica forzosamente el objeto dentro de una de las clases que conoce, a veces incluso con un nivel de confianza engañosamente alto basándose únicamente en colores, fondos o texturas similares. Para evitar esto en una aplicación real, es necesario fijar un umbral mínimo de confianza (como el implementado en la interfaz web) o entrenar una clase explícita de "fondo/vacío".
